```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('sentiment_analysis_sample_100000.csv')
```

```python
df
```

```python
x = df['sentence']
y = df['sentiment']
```

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

tfidf = TfidfVectorizer()
x_train_tfidf = tfidf.fit_transform(x_train)

x_test_tfidf = tfidf.transform(x_test)

print("Training Feature:", x_train_tfidf)
print("Testing feature:", x_test_tfidf)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()

model.fit(x_train_tfidf, y_train)

print("Training Completed!")
```

```python
from sklearn.metrics import accuracy_score

y_pred = model.predict(x_test_tfidf)
print(accuracy_score(y_test, y_pred))
```

```python
testcase = ["You are good"]
testcase_tfidf = tfidf.transform(testcase)
response = model.predict(testcase_tfidf)
print(response)
```

```python
testcase = ["What a confusing experience with the phone"]
testcase_tfidf = tfidf.transform(testcase)
response = model.predict(testcase_tfidf)
print(response)
```
