```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.metrics import accuracy_score, classification_report
from sklearn.model_selection import cross_val_score, train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler

# 1. Load data
df = pd.read_csv('/content/student_clustering.csv')
X = df[['cgpa', 'iq']]
```

```python
scaler_init = StandardScaler()
y = KMeans(n_clusters=4, random_state=42, n_init=10).fit_predict(
    scaler_init.fit_transform(X)
)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
knn = KNeighborsClassifier(n_neighbors=5)
cv_acc = cross_val_score(knn, X_train_scaled, y_train, cv=5).mean()
```

```python
knn = KNeighborsClassifier(n_neighbors=5)
cv_acc = cross_val_score(knn, X_train_scaled, y_train, cv=5).mean()

knn.fit(X_train_scaled, y_train)
y_pred = knn.predict(X_test_scaled)
test_acc = accuracy_score(y_test, y_pred)

print(f'5-Fold CV Accuracy: {cv_acc * 100:.2f}%')
print(f'Test Accuracy (K=5): {test_acc * 100:.2f}%')
print(classification_report(y_test, y_pred))
```
